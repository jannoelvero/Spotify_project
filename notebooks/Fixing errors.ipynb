```python
from pathlib import Path
import pandas as pd

PROJECT_ROOT = Path.home() / "Documents" / "spotify_project"
CLEAN_DATA = PROJECT_ROOT / "data" / "clean"

print(CLEAN_DATA)
```

```python
df_sql = pd.read_csv(
    CLEAN_DATA / "spotify_final.csv"
)

print(df_sql.shape)
```

```python
from sqlalchemy import create_engine
import getpass

password = getpass.getpass("Enter MySQL root password: ")

engine = create_engine(
    f"mysql+pymysql://root:{password}@localhost:3306/spotify_db"
)

print("Engine created successfully")
```

```python
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
import getpass

password = getpass.getpass("Enter MySQL root password: ")

connection_url = URL.create(
    "mysql+pymysql",
    username="root",
    password=password,
    host="localhost",
    port=3306,
    database="spotify_db"
)

engine = create_engine(connection_url)

print("Engine created successfully")
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(text("SELECT DATABASE();"))
    print("Connected database:", result.scalar())
```

```python
df_sql.to_sql(
    name="spotify_tracks",
    con=engine,
    if_exists="append",
    index=False
)
```

```python
from sqlalchemy import text

with engine.connect() as connection:
    result = connection.execute(
        text("SELECT COUNT(*) FROM spotify_tracks;")
    )
    print("Total rows:", result.scalar())
```
